```python
from sklearn import datasets # for importing the examples
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt # for visualisation
from sklearn.model_selection import train_test_split # for spliting data
from sklearn.ensemble import RandomForestClassifier # for random forest classifier
import seaborn as sns
```

```python
penguins = sns.load_dataset("penguins")
penguins.info()
```

```python
penguins.head()
```

```python
x = penguins.drop(columns=['species', 'sex','island'])
y = penguins['species']
```

```python
penguins.species.value_counts()
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(x,y,random_state=42, test_size=0.2)
```

```python
rfc = RandomForestClassifier(n_estimators=100, max_samples=0.5, random_state=42,oob_score=True)
```

```python
rfc.fit(X_train, Y_train)
```

```python
rfc.oob_score_
```

```python
y_pred = rfc.predict(X_test)
```

```python

from sklearn.metrics import accuracy_score
y_pred = rfc.predict(X_test)
accuracy_score(Y_test, y_pred) # Instead of r2 in regression , we use accuracy_score
```

```python


from sklearn.metrics import confusion_matrix
cm = confusion_matrix(Y_test, y_pred)
print(cm)
```

```python


sns.heatmap(cm, annot=True, cmap='Blues')
plt.show()
```
